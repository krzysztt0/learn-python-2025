# Recap: Control flow

````python
# some lines of code executed line-by-line
def square(number):
    return number**2

x = 1
y = 5
z = x + y

print(z)

if z > 3:
    print("z > 3")
elif z == 3:
    print("z == 3")
else: 
    print("z < 3")

print("end of if-elif-else")

print(square(z))

for i in range(z):
    print(f"i is {i}")

print("end of for-loop")

while z > 0:
    z -= 1
    print(f"z is {z}")

print("end of while-loop")

print("Done!")
````
