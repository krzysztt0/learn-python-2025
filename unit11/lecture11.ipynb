# Part 2 - Recap: Control flow

````python
# some lines of code executed line-by-line
def square(number):
    return number**2

x = 1
y = 5
z = x + y

print(z)

if z > 3:
    print("z > 3")
elif z == 3:
    print("z == 3")
else: 
    print("z < 3")

print("end of if-elif-else")

print(square(z))

for i in range(z):
    print(f"i is {i}")

print("end of for-loop")

while z > 0:
    z -= 1
    print(f"z is {z}")

print("end of while-loop")

print("Done!")
````

# pandas recap with the spotify2018 data set

* import pandas: `import pandas as pd`
* read in a csv file `pd.read_csv()`
* Boolean indexing
* combine several filtering conditions with `&`, `|`
* retrieve values `.values[0]`

````python
# import pandas
import pandas as pd
````

````python
# read in the csv "2018.csv"
df = pd.read_csv("2018.csv")
````

````python
# view the head
df.head()
````

````python
# boolean indexing: only those rows where "genre"=="pop"
df["genre"] == "pop" # this gives us a Series of booleans
````

````python
# boolean indexing: only those rows where "genre"=="pop"
df[ df["genre"] == "pop" ] # here we use the Series of booleans to filter the rows
````

````python
# boolean indexing: only those rows where "genre"=="hip hop" AND "explicit"==False
df[ (df["genre"]=="hip hop") & (df["explicit"]==False) ] # here we combine 2 conditions with "&"
````

````python
# boolean indexing: only those rows where "explicit"==False, AND ("genre"=="rock" OR "genre"=="pop") (use | for "or" in pandas)
df[ (df["explicit"]==False) & ((df["genre"]=="rock") | (df["genre"]=="pop"))  ]
# df[ (df["explicit"]==False) & df["genre"].isin(["rock", "pop"])  ] # same as above, but using the `.isin()` method
````

````python
# retrieve a certain value:
# what is the only MGMT song in this data set?
# save the value in the "song" column (a string) to a variable
df[df["artist"]=="MGMT"]#["song"]#.values[0]
````

````python
# remember to .copy() before modifying!
hiphop = df[df["genre"]=="hip hop"].copy() # !!!
hiphop
````

# `random` module recap

````python
import random
````

````python
# choose 1 random element from a non-empty sequence
random.choice(range(1,7))
````

````python
# choose k random elements (with replacement!) from a population
# i.e. it could happen that the same element is chosen repeatedly
random.choices(
    population = range(1,11), # numbers from 1 to 10
    k = 5 # how many elements to return? 
)
````

````python
# choose k random elements (without replacement!) from a population
random.sample(
    population = range(1,11), # numbers from 1 to 10
    k = 5 # how many elements to return?
)
````

````python
# get a random number in the range from a to b
# ("sampling from a uniform distribution")
random.uniform(a=0, b=1)
````
