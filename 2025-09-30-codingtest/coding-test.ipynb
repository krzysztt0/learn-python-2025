# Task 1: Password checker

1. A password has to have at least 12 characters.

3. A password has to contain at least one lowercase letter **and** at least one uppercase letter. ## Task 1a (1 point): `check_length`

Write the function `check_length(password)` that takes as input a string paramter, `password`. It returns `True` if the password is at least 12 characters long, and it returns `False` otherwise. (That means that only Rule 1 from above is implemented.)

## Task 1b (1 point): `is_valid`

Write the function `is_valid(password)` that takes as input a string parameter, `password`. It returns `True` if the password fulfills **all 3 rules** above, and `False` otherwise. 


````python
# YOUR SOLUTION TO TASK 1A
def check_length(password):
    if len(str(password)) >= 12:
        return True
    else:
        return False
    

print(check_length("thispasswordislongenough"))
````

````python
# YOUR SOLUTION TO TASK 1B
def is_valid(password):
    
    p_str = str(password)
    if len(p_str) < 12:
        return False
    

    #check no repeated characters
   

    p_list = list(p_str)

    p_set = set(p_list)

    if len(p_set) == 1:
        return False


    #check lower case and upper case characters     
    for i in range(len(p_str)):
        if p_str.islower() and p_str.isupper():
            return True 
        else:
            return False




````
