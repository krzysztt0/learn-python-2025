# Section A: List & loop concepts

**Key Concepts**: Lists, for loops, while loops, indexing, range()

```python
# TASK 1: Count negative numbers using for loop
# NOTE: This task demonstrates basic for loop with range() and list indexing

L = [-1,2,-3,4,12,12]

def count(L):
    """Count negative numbers in a list using for loop"""
    cnt = 0
    for i in range(len(L)):  # Use len(L) instead of global variable for better practice
        if L[i] < 0:
            cnt += 1
    return cnt  

count(L)
```

```python
# TASK 2: Sum all elements using while loop
# NOTE: While loops require manual index management (increment i!)
# COMMON MISTAKE: Forgetting to increment i causes infinite loop!

def sum_list(L):  # Renamed to avoid conflict with built-in sum()
    """Sum all elements in a list using while loop"""
    res = 0
    i = 0
    while i < len(L):  # Condition: continue while i is less than list length
        res += L[i]    # Add current element to result
        i += 1         # CRITICAL: increment i to avoid infinite loop!
    return res

print(sum_list(L))
```

```python
# TASK 3: Count occurrences of specific value (12) using while loop
# NOTE: Similar to Task 2 but with conditional counting

def count_12(L):  # Renamed for clarity
    """Count how many times 12 appears in the list"""
    result = 0
    i = 0
    while i < len(L):
        if L[i] == 12:     # Check if current element equals 12
            result += 1    # Increment counter only if condition is true
        i += 1             # Always increment i (inside or outside if doesn't matter)
    return result

print(count_12(L))
```

```python
# TASK 3: Danish Royal Succession Pattern
# DIFFICULTY: ★★★★☆ (Hard) - String parsing + complex logic

def next_king(name):
    """
    Returns the next Danish king following these rules:
    1. Christian ↔ Frederick (names alternate)
    2. Frederick → Christian: number increases by 1
    3. Christian → Frederick: number stays the same
    """
    
    parts = name.strip().split()  # Split "Christian 7" into ["Christian", "7"]
    
    if len(parts) == 1:
        # Handle case with no number (half points solution)
        if parts[0] == "Christian":
            return "Frederick"
        elif parts[0] == "Frederick":
            return "Christian"
        else:
            return "Unknown king name"
    
    elif len(parts) == 2:
        # Handle case with name and number
        king_name = parts[0]
        number = int(parts[1])
        
        if king_name == "Christian":
            # Christian -> Frederick (same number)
            return f"Frederick {number}"
        elif king_name == "Frederick":
            # Frederick -> Christian (number increases by 1)
            return f"Christian {number + 1}"
        else:
            return "Unknown king name"
    
    else:
        return "Invalid input format"
```

```python
# TASK 4: Course Search System with Autocompletion
# DIFFICULTY: ★★★★☆ (Hard) - Complex interactive program with multiple conditions

courses = ["introduction to programming", "introduction to machine learning", "advanced statistics"]

# Step 1: Basic course search
def course_search_basic():
    """Interactive program that repeatedly asks for course input."""
    while True:
        course = input("Please provide a course: ")
        
        if course == "":
            break
            
        # Check if course exists
        if course in courses:
            print("This course exists at ITU.")
        else:
            print("No such course exists at ITU.")

# Step 2: Advanced course search with autocompletion
def course_search_autocomplete():
    """Advanced search with autocompletion based on prefix matching."""
    while True:
        course = input("Please provide a course: ")
        
        if course == "":  # Empty input → quit
            break
            
        # Find all courses that start with input string (prefix matching)
        matching_courses = [c for c in courses if c.startswith(course)]
        
        if len(matching_courses) == 0:
            print("No courses at ITU start like this.")
            break  # Quit after no matches
        elif len(matching_courses) == 1:
            print(f"The course you mean is {matching_courses[0]}")
            break  # Quit after finding unique match
        else:
            # Multiple matches - show them and continue asking
            print("Multiple courses at ITU start like this:")
            for c in matching_courses:
                print(c)
            # Loop continues for more specific input
```

**Key Concepts**: File I/O, string parsing, dictionaries, data aggregation
**File Formats**: 
- trips.txt: "Country, City" (comma-space separated)
- delays.txt: "Start Destination Delay" (space separated)

```python
# TASK 1: Travel Data Analysis
# DIFFICULTY: ★★★☆☆ (Medium) - File processing + dictionary operations

# Step 1: Check if specific city-country combination exists
def has_visited(filename, country, city):
    """Search through travel file to find if Rasmus visited a specific city in a country."""
    with open(filename, 'r') as f:
        for line in f:
            visited_country, visited_city = line.strip().split(', ')
            if visited_country == country and visited_city == city:
                return True
        return False

# Step 2: Function to find the most visited country
def most_visited(filename):
    """Find and print the country Rasmus visited most frequently."""
    country_count = {}
    with open(filename, 'r') as f:
        for line in f:
            country, city = line.strip().split(', ')
            if country in country_count:
                country_count[country] += 1
            else:
                country_count[country] = 1
    
    # Find the country with maximum visits
    max_visits = max(country_count.values())
    for country, visits in country_count.items():
        if visits == max_visits:
            print(country)
            return
```

```python
# TASK 2: Train Delay Analysis
# DIFFICULTY: ★★★★☆ (Hard) - Complex file parsing + mathematical operations

# Step 1: Find and analyze specific route delays
def print_trips(filename, start, destination):
    """Find all trips from start to destination and calculate average delay."""
    delays = []
    trips_found = []
    
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 3:
                trip_start = parts[0]
                trip_destination = parts[1]
                delay = int(parts[2])
                
                if trip_start == start and trip_destination == destination:
                    trips_found.append(line.strip())
                    delays.append(delay)
    
    if trips_found:
        for trip in trips_found:
            print(trip)
        average_delay = sum(delays) // len(delays)
        print(f"Average delay: {average_delay} minutes")
    else:
        print("No trip found.")

# Step 2: Find connection with maximum average delay
def find_maximum_average_delay(filename):
    """Return the connection with the maximum average delay among all trips."""
    connections = {}  # Dictionary to store {(start, destination): [delays]}
    
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 3:
                start = parts[0]
                destination = parts[1]
                delay = int(parts[2])
                
                connection = (start, destination)
                if connection in connections:
                    connections[connection].append(delay)
                else:
                    connections[connection] = [delay]
    
    max_average = 0
    max_connection = None
    
    for connection, delays in connections.items():
        average = sum(delays) // len(delays)
        if average > max_average:
            max_average = average
            max_connection = connection
    
    return f"{max_connection[0]} {max_connection[1]}"
```
