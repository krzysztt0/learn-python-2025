T1.1 :: Fix the given function, that returns only the strings in list L that are lower-case

```python
def collect_lower_case(L): 
    
    n = len(L) 
    i = 0

    # create a new list to store lower-case strings
    new_list = [] 
    
    # iterate through input list
    while i < n: 

        # update new list when a lower-case string is found
        if L[i].islower(): 
            new_list.append(L[i]) 
        
        # update i to proceed along the input list values
        i = i + 1 

    return new_list
```

## Task 2   
For a valid pin n, the following constraints hold:
1. The pin has 4 digits
2. The pin does not consist of one unique repeated digit (1111, 22222, ..., 9999)
3. The pin does not consist of an ascending or descending order of successive digits (e.g. 1234, 8765, ...)


T2.1 :: Implement a function that checks exclusively that the given pin n has 4 digits, returning True when that is the case, and False otherwise

```python
def is_valid_naive(n):

    # to check the length of our integer input, 
    # convert it to a string first  
    return len(str(n)) == 4
```

T2.2 :: Implement a function that returns True if the given pin n doesn't violate any of the 3 constraints, and False otherwise

```python
def is_valid(n):

    # check 1.
    n_str = str(n)
    if len(n_str) != 4:
        return False
    

    # check 2.

    # create a list with each digit as an element
    n_list = list(n_str)

    # eliminate duplicates
    n_set = set(n_list)

    # if there is only one unique digit, 
    # the pin violates constraint 2.
    if len(n_set) == 1:
        return False


    # check 3.

    # ascending sequences

    # set up a flag variable (True means constraint 3. is violated)
    x = True

    # iterate through the STRINGS corresponding to digits
    for i in range(len(n_list)-1):

        # as soon as a pair of INT is not in a successive ascending sequence,
        # we can state that constraint 3. is met, and we update the flag variable
        if int(n_list[i]) != int(n_list[i+1]) + 1:
            x = False
            break
    
    # if it is True that condition 3. doesn't hold, the pin is not valid
    if x:
        return False


    # descending sequences

    # set up a flag variable (True means constraint 3. is violated)
    y = True

    # iterate through the STRINGS corresponding to digits
    for i in range(len(n_list)-1):

        # as soon as a pair of INT is not in a successive descending sequence,
        # we can state that constraint 3. is met, and we update the flag variable
        if int(n_list[i]) != int(n_list[i+1]) - 1:
            y = False
            break
    
    # if it is True that condition 3. doesn't hold, the pin is not valid
    if y:
        return False
    
    # if the function hasn't returned False yet, 
    # it means that the pin is valid
    return True
```

### Task 3  
A student graduates if they manage to pass five exams (A, B, C, D, E), graded with the Danish system (2, 4, 7, 10, 12 are the passing grades).  
In the file 'grades.txt', each line represents an exam result, in the following format: " 'Name', 'Exam', 'Grade' ".  
Assume each student has a unique name, and that each student cannot retake an exam that they previously passed.

T3.1 :: Create a function that, for a given filename fn and person's name, prints the results of all exams sustained by that person

```python
def display_results(fn, name):
    
    with open(fn) as f:
        for l in f:
            l = l.strip("\n")  # not fundamental

            # split the line using the specified format of "'Name', 'Exam', 'Grade'":
            # this will give us a list with ['Name', 'Exam', 'Grade']
            l_list = l.split(", ")

            # check that the name matches the one that is being requested
            l_name = l_list[0]
            if l_name == name:
                print(l)
```

T3.2 :: Create a function that for a given filename fn, prints the names of the people who graduated

```python
def print_finished(fn):

    # create a dictionary to store students and their exam situation
    dict_students = {}

    with open(fn) as f:
        for l in f:
            l = l.strip("\n")  # not fundamental

            # split the line using the specified format of "'Name', 'Exam', 'Grade'":
            # this will give us a list with ['Name', 'Exam', 'Grade']
            l_list = l.split(", ")
            
            # store Name, Exam, and Grade into variables
            l_name = l_list[0]
            l_exam = l_list[1]
            l_grade = int(l_list[2])

            # .setdefault() method: if the given key doesn't exist in the dictionary, add it with the associated value,
            # if it does exist, do nothing.
            # In this case, we add each student's name as a key (remember we assumed names are unique), and a dictionary
            # mapping each exam to a grade as a value, where 'False' is our value for when a student has yet to take that exam
            dict_students.setdefault(l_name, {"A" : False, "B" : False, "C" : False, "D" : False, "E" : False})

            # obtain the current result in the given exam
            result = dict_students[l_name][l_exam]

            # if it is a passing grade, store it
            if l_grade >= 2 and result <= 0:  # the 'result <= 0' check is to prevent updating the grade for an exam that
                                              # was already passed, since the rules state you should not be able to re-attempt
                                              # an exam once you have a grade greater than 0. This makes the program more robust
                                              # to errors in the data file, even though in this case there are none, so it is not
                                              # a must-have.
                dict_students[l_name][l_exam] = l_grade

    # now iterate to the complete dictionary
    for k, v in dict_students.items():

        # flag variable to indicate wether the student k graduated
        graduated = True

        # check the grades, if ANY of them is not a passing grade,
        # the student did not yet graduate and the flag variable is updated
        for grade in v.values():
            if grade <= 0:
                graduated = False
                break
        
        # print the name only if it has all 5 passing grades required
        if graduated:
            print(k)
```

T3.3 :: Create a function that given a filename fn, prints the name and average grade of the student who graduated with the best average grade.  
You can assume there are no students with the same average.

```python
# average accounting exclusively for passing grades
def print_best_student_PassGrades(fn):

    # create a dictionary to store students and their exam situation
    dict_student = {}  # name : (n. of exams passed, total grade)
    with open(fn) as f:
        for l in f:
            l = l.strip("\n")  # not fundamental

            # split the line using the specified format of "'Name', 'Exam', 'Grade'":
            # this will give us a list with ['Name', 'Exam', 'Grade']
            l_list = l.split(", ")
            
            # store Name, Exam, and Grade into variables
            l_name = l_list[0]
            l_exam = l_list[1]
            l_grade = int(l_list[2])

            # .setdefault() method: if the given key doesn't exist in the dictionary, add it with the associated value,
            # if it does exist, do nothing.
            # In this case, we store student name, and a list that stores number of exams passed, and total grade
            dict_student.setdefault(l_name, [0, 0])

            # only if it is a passing grade, we update both the number of exams passed, and the total grade
            if l_grade >= 2:
                dict_student[l_name][0] += 1
                dict_student[l_name][1] += l_grade
    
    # set up a new dictionary to store average grades
    dict_averages = {}
    for k, v in dict_student.items():

        # only if the student graduated (that is, the number of exams passed is equal to 5)
        # we compute and store their mean grade
        if v[0] == 5:
            dict_averages[k] = v[1] / 5

    # sort the dictionary according to the averages, 
    # with the best average being the first entry in the resulting list
    averages = sorted(list(dict_averages.items()), key= lambda x: x[1], reverse= True)
    result = averages[0]

    # print name and average of the person who graduated with the highest average grade
    print(f"{result[0]}, {result[1]}")


# average accounting for all grades, passing and non passing
def print_best_student_AllGrades(fn):

    # create a dictionary to store students and their exam situation
    dict_student = {}  # name : (n. of exams passed, total n. of exams, total grade)
    with open(fn) as f:
        for l in f:
            l = l.strip("\n")  # not fundamental

            # split the line using the specified format of "'Name', 'Exam', 'Grade'":
            # this will give us a list with ['Name', 'Exam', 'Grade']
            l_list = l.split(", ")
            
            # store Name, Exam, and Grade into variables
            l_name = l_list[0]
            l_exam = l_list[1]
            l_grade = int(l_list[2])

            # .setdefault() method: if the given key doesn't exist in the dictionary, add it with the associated value,
            # if it does exist, do nothing.
            # In this case, we store student name, and a list that stores number of exams passed, total number of exams, and total grade
            dict_student.setdefault(l_name, [0, 0, 0])

            # only if it is a passing grade, we update the number of exams passed
            if l_grade >= 2:
                dict_student[l_name][0] += 1
            
            # no matter the grade, we update the total number of exams, and total grade
            dict_student[l_name][1] += 1
            dict_student[l_name][2] += l_grade
    
    # set up a new dictionary to store average grades
    dict_averages = {}
    for k, v in dict_student.items():

        # only if the student graduated (that is, the number of exams passed is equal to 5)
        # we compute and store their mean grade
        if v[0] == 5:
            dict_averages[k] = v[2] / v[1]  # recall: v[2] is the total grade, v[1] the total number of exams

    # sort the dictionary according to the averages, 
    # with the best average being the first entry in the resulting list
    averages = sorted(list(dict_averages.items()), key= lambda x: x[1], reverse= True)
    result = averages[0]

    # print name and average of the person who graduated with the highest average grade
    print(f"{result[0]}, {result[1]}")
```
