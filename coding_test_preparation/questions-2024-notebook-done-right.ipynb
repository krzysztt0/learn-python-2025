```python
# TASK 1: YOUR CODE HERE
# Step 1:
def check_ta_count(rooms, tas):
    if tas >= rooms:
        print("Thank you very much for such a generous support.")
    else:
        print("Unfortunately, I need more TAs to run the exercise sessions.")

# Step 2:
def assign_tas(rooms, tas):
    tas_left = tas
    rooms_left = rooms
    while(tas_left > 0):
        if tas_left >= rooms_left:
            print(f"TA {tas_left} goes to room {rooms_left}.")
        else:
            rooms_left -= 1
            print(f"TA {tas_left} goes to room {rooms_left}.")
        tas_left -= 1
assign_tas(5,10)
```

# Section C: Working with read-in data

```python
# TASK 1: YOUR CODE HERE
# Step 1:
def has_visited(filename, country, city):
    with open(filename, "r") as file:
        visits = file.read()
    if f"{country}; {city}" in visits or f"{country}, {city}" in visits:
        return(True)
    return(False)

print(has_visited("trips.txt", "Denmark", "Aarhus"))

# Step 2:

def most_visited(filename):
    with open(filename, "r") as file:
        visits = file.read()
    visitcount = dict()
    locations = visits.split()
    for location in locations:
        if location in visitcount:
            visitcount[location] += 1
        else:
            visitcount[location] = 1
    print(max(visitcount, key = visitcount.get))
```
