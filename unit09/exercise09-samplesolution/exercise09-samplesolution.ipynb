# Task 2: $\pi$, but faster

In lecture 09, we already approximated $\pi$ by Monte Carlo simulation with a simple for-loop, and using the built-in `random` module. 

**2a.** Recreate (copy-paste?) the $\pi$ approximation from the lecture notebook.

**2b.** Implement the $\pi$ approximation by Monte Carlo simulation one more time, but this time using `numpy` functions. (see below - detailed numpy introduction will follow soon!)

**2c.** Compute and compare runtimes for different $N$ (number of trials) for the Monte Carlo simulations with (2a) vs. without (2b) numpy. To make runtime comparison easier, wrap both implementations into function definitions. Then use `time.time_ns()` to compare runtimes for the following sizes of N:

```python
[10**6, 10**7, 5*10**7, 10**8, 5*10**8]
```

(Note that for N ~ 10**8 you might need to wait for a couple of minutes for the code to run.)

***

`numpy` hints:
* instead of `random.uniform()`, use [`numpy.random.uniform()`](https://numpy.org/doc/2.3/reference/random/generated/numpy.random.uniform.html).
* if you used the built-in function `sum()`: replace it with the `.sum()` method instead
* To nerd out a bit more on numpy runtime (and the limits for speedup with numpy), check for example: https://pythonspeed.com/articles/vectorization-python-alternatives/ 

````python
# 2a - pi approximation with for loop and random module:

in_quartercircle = 0
total_points = 10**6
for _ in range(total_points):
    x = random.uniform(0,1)
    y = random.uniform(0,1)
    if x**2 + y**2 < 1:
        in_quartercircle += 1
pi_approx = 4 * in_quartercircle / total_points
print(f"approximated pi: {pi_approx}" )
````

````python
# 2b - pi approximation with numpy (vectorized):

total_points = 10**6

# generate N random trials at once:
xs = np.random.uniform(0,1, size=total_points)
ys = np.random.uniform(0,1, size=total_points)
in_quartercircle = (xs**2 + ys**2 < 1).sum() 
pi_approx = 4 *  in_quartercircle / total_points
print(f"approximated pi: {pi_approx}" )
````

````python
# 2c: comparing runtime

def pi_approx_withoutnumpy(N):
    '''
    approximate pi through a Monte Carlo simulation with N points
    without using numpy
    '''
    in_quartercircle = 0
    for _ in range(N):
        x = random.uniform(0,1)
        y = random.uniform(0,1)
        if x**2 + y**2 < 1:
            in_quartercircle += 1
    pi_approx = 4 * in_quartercircle / N
    return pi_approx

def pi_approx_withnumpy(N):
    '''
    approximate pi through a Monte Carlo simulation with N points
    without using numpy
    '''
    xs = np.random.uniform(0,1, size=N)
    ys = np.random.uniform(0,1, size=N)
    pi_approx = 4 * (xs**2 + ys**2 < 1).sum() / N
    return pi_approx
````

````python
for N in [10**6, 10**7, 5*10**7, 10**8, 5*10**8]:
    time1 = time.time_ns()
    pi_approx_withoutnumpy(N)
    time2 = time.time_ns()
    time_without_numpy = time2 - time1

    time3 = time.time_ns()
    pi_approx_withnumpy(N)
    time4 = time.time_ns()
    time_with_numpy = time4 - time3

    x = round(time_without_numpy / time_with_numpy)

    print(f"For {N} trials, numpy-based simulation is ~ {x} times faster")
````
